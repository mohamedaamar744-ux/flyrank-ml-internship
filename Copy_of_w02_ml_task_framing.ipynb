## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

**Why ML beats a fixed rule**

A fixed rule could prioritize every page with a `down` trend, but this uses only one signal and treats all declining pages the same. A scoring model can combine multiple measurable signals about content performance and assign a score that helps rank pages for review.

This is useful because the team has limited time and resources, so it needs to decide which pages to review first. The model's output can support this prioritization instead of relying on a single fixed condition.
